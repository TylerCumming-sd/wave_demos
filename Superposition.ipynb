```python
#Program: Superposition.ipynb
#Purpose: To provide a visual illustration of superposition of waves and the impact of varying different parameters
#Author: Tyler Cumming
#Date:03/05/2025

#Importing relevant libraries
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interactive

#Parameter definitions
#A=Amplitude
#k=Wavenumber
#w=Angular Frequency
#t=Time
#phi=Phase Shift

#Function for plotting waves
def plot_wave(A_1=1,A_2=1,k_1=1,k_2=1,w_1=1,w_2=1,t=0,x_max=10,phi_1=0,phi_2=0):
    x=np.linspace(0,x_max,1000)
    y_1=A_1*np.sin(k_1*x-w_1*t+phi_1)
    y_2=A_2*np.sin(k_2*x-w_2*t+phi_2)
    y_sum=y_1+y_2
    plt.title('Classical Wave with Differing Parameters')
    plt.xlabel('Position (x)')
    plt.ylabel('Amplitude (y)')
    plt.grid(True)
    plt.plot(x, y_1, label=f'Wave 1: A={A_1}, k={k_1}, w={w_1}, φ={phi_1}',color='red')
    plt.plot(x, y_2, label=f'Wave 2: A={A_2}, k={k_2}, w={w_2}, φ={phi_2}',color='blue')
    plt.plot(x, y_sum, label='Sum of Waves', color='black', linestyle='dashed')
    plt.legend(loc='upper right')


#Change Parameters
# Create sliders for interactive input
A_1_slider = widgets.FloatSlider(value=1, min=0, max=5, step=0.1, description='Amplitude (A):')
k_1_slider = widgets.FloatSlider(value=1, min=0, max=5, step=0.1, description='Wave Number (k):')
w_1_slider = widgets.FloatSlider(value=1, min=0, max=5, step=0.1, description='Angular Frequency (w):')
phi_1_slider = widgets.FloatSlider(value=0,min=-np.pi,max=np.pi,step=0.1,description='Phase Shift($\phi)')
A_2_slider = widgets.FloatSlider(value=1, min=0, max=5, step=0.1, description='Amplitude (A):')
k_2_slider = widgets.FloatSlider(value=1, min=0, max=5, step=0.1, description='Wave Number (k):')
w_2_slider = widgets.FloatSlider(value=1, min=0, max=5, step=0.1, description='Angular Frequency (w):')
phi_2_slider = widgets.FloatSlider(value=0,min=-np.pi,max=np.pi,step=0.1,description='Phase Shift($\phi)')
t_slider = widgets.FloatSlider(value=0, min=0, max=120, step=1, description='Time (t):')
x_max_slider = widgets.FloatSlider(value=10, min=0, max=20, step=0.5, description='X Max')


# Create an interactive plot
interactive_plot = interactive(plot_wave, A_1=A_1_slider,A_2=A_2_slider, k_1=k_1_slider, k_2=k_2_slider, w_1=w_1_slider,w_2=w_2_slider, t=t_slider, x_max=x_max_slider,phi_1=phi_1_slider,phi_2=phi_2_slider)

# Display the interactive widgets
display(interactive_plot)
```
