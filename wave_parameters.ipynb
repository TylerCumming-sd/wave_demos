```python
#Simple program to model classical waves

#importing relevant libraries
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interactive

#parameter definitions
#A=Amplitude
#k=Wavenumber
#w=Angular Frequency
#t=Time
#phi=Phase Shift

#function for plotting wave
def plot_wave(A=1,k=1,w=1,t=0,x_max=10,phi=0):
    x=np.linspace(0,x_max,1000)
    y=A*np.sin(k*x-w*t+phi)
    v_p=w/k
    plt.title('Classical Wave with Differing Parameters')
    plt.xlabel('Position (x)')
    plt.ylabel('Amplitude (y)')
    plt.grid(True)
    plt.plot(x,y)
    print(f'The phase velocity of the wave is {v_p:.2f}')

#Change Parameters
# Create sliders for interactive input
A_slider = widgets.FloatSlider(value=1, min=0, max=5, step=0.1, description='Amplitude (A):')
k_slider = widgets.FloatSlider(value=1, min=0, max=5, step=0.1, description='Wave Number (k):')
w_slider = widgets.FloatSlider(value=1, min=0, max=5, step=0.1, description='Angular Frequency (w):')
t_slider = widgets.FloatSlider(value=0, min=0, max=120, step=1, description='Time (t):')
x_max_slider = widgets.FloatSlider(value=10, min=0, max=20, step=0.5, description='X Max')
phi_slider = widgets.FloatSlider(value=0,min=-np.pi,max=np.pi,step=0.1,description='Phase Shift($\phi)')

# Create an interactive plot
interactive_plot = interactive(plot_wave, A=A_slider, k=k_slider, w=w_slider, t=t_slider, x_max=x_max_slider,phi=phi_slider)

# Display the interactive widgets
display(interactive_plot)
```
